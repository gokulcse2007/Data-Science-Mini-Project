```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python
from google.colab import files

uploaded = files.upload()
```

```python
df = pd.read_csv('StudentsPerformance.csv')

print("Dataset loaded successfully!")

display(df.head())
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
df = df.drop_duplicates()

df = df.dropna()

print("Data cleaning completed!")

print("Updated Dataset Shape:")
print(df.shape)
```

```python
df['average_score'] = (
    df['math score'] +
    df['reading score'] +
    df['writing score']
) / 3

print("Average Scores Added Successfully!")

display(df.head())
```

```python
print("Average Mathematics Score:",
      df['math score'].mean())

print("Average Reading Score:",
      df['reading score'].mean())

print("Average Writing Score:",
      df['writing score'].mean())

print("Overall Average Score:",
      df['average_score'].mean())
```

```python
subjects = ['Math', 'Reading', 'Writing']

averages = [
    df['math score'].mean(),
    df['reading score'].mean(),
    df['writing score'].mean()
]

plt.figure(figsize=(8,5))

plt.bar(subjects, averages)

plt.title("Average Scores by Subject")
plt.xlabel("Subjects")
plt.ylabel("Average Score")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df['math score'], bins=10, edgecolor='black')

plt.title("Distribution of Mathematics Scores")
plt.xlabel("Math Score")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x='math score',
    y='reading score'
)

plt.title("Math Score vs Reading Score")
plt.xlabel("Math Score")
plt.ylabel("Reading Score")

plt.show()
```

```python
preparation = df['test preparation course'].value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    preparation.values,
    labels=preparation.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Test Preparation Course Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=df,
    x='test preparation course',
    y='average_score'
)

plt.title("Test Preparation vs Average Score")
plt.xlabel("Test Preparation")
plt.ylabel("Average Score")

plt.show()
```

```python
plt.figure(figsize=(8,6))

correlation = df[
    ['math score', 'reading score', 'writing score']
].corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm'
)

plt.title("Student Score Correlation")

plt.show()
```

```python
print("STUDENT PERFORMANCE ANALYSIS")
print("--------------------------------")

print("Total Students:", len(df))

print("Average Math Score:",
      round(df['math score'].mean(), 2))

print("Average Reading Score:",
      round(df['reading score'].mean(), 2))

print("Average Writing Score:",
      round(df['writing score'].mean(), 2))

print("Highest Average Score:",
      round(df['average_score'].max(), 2))

print("Lowest Average Score:",
      round(df['average_score'].min(), 2))

print("\nBest Performing Subject:")

subject_averages = df[
    ['math score', 'reading score', 'writing score']
].mean()

print(subject_averages.idxmax())

print("\nAverage Score Based on Test Preparation:")

print(
    df.groupby('test preparation course')['average_score'].mean()
)
```
